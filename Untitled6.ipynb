```python
import os
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, random_split

data_dir = 'dataset'
print("Conteúdo da pasta dataset:", os.listdir(data_dir))
```

```python
image_size = 224

train_transforms = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.RandomHorizontalFlip(p=0.5),          # Inversão horizontal aleatória
    transforms.RandomRotation(degrees=15),           # Pequena rotação nas imagens
    transforms.ColorJitter(brightness=0.2, contrast=0.2), # Variação de brilho/contraste
    transforms.ToTensor(),                           # Converte para Tensor do PyTorch
    transforms.Normalize([0.485, 0.456, 0.406],      # Normalização padrão do ImageNet
                         [0.229, 0.224, 0.225])
])

# Transformação padrão (sem aumento) para Validação e Teste
val_test_transforms = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

print("Transformações e Data Augmentation configurados com sucesso!")
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from torch.utils.data import Subset


full_dataset_train = datasets.ImageFolder(data_dir, transform=train_transforms)
full_dataset_val_test = datasets.ImageFolder(data_dir, transform=val_test_transforms)


targets = full_dataset_train.targets


train_idx, temp_idx, _, temp_targets = train_test_split(
    np.arange(len(targets)),
    targets,
    test_size=0.3,
    stratify=targets,
    random_state=42
)


val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.5,
    stratify=temp_targets,
    random_state=42
)

train_dataset = Subset(full_dataset_train, train_idx)
val_dataset = Subset(full_dataset_val_test, val_idx)
test_dataset = Subset(full_dataset_val_test, test_idx)


batch_size = 32

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False) # Teste só será usado no final

print(f"Total de imagens no Treino: {len(train_dataset)}")
print(f"Total de imagens na Validação: {len(val_dataset)}")
print(f"Total de imagens no Teste: {len(test_dataset)}")
```

```python
import os
import torch.nn as nn


pasta_correta = 'dataset'
for root, dirs, files in os.walk('dataset'):
    if len(dirs) > 5:  # O weather-dataset tem cerca de 11 classes
        pasta_correta = root
        break

print(f"A pasta correta com as imagens é: {pasta_correta}")


full_dataset_train = datasets.ImageFolder(pasta_correta, transform=train_transforms)
full_dataset_val_test = datasets.ImageFolder(pasta_correta, transform=val_test_transforms)


class_names = full_dataset_train.classes
num_classes = len(class_names)
print(f"Classes detetadas ({num_classes}): {class_names}")


targets = full_dataset_train.targets
train_idx, temp_idx, _, temp_targets = train_test_split(
    np.arange(len(targets)), targets, test_size=0.3, stratify=targets, random_state=42
)
val_idx, test_idx = train_test_split(
    temp_idx, test_size=0.5, stratify=temp_targets, random_state=42
)

train_dataset = Subset(full_dataset_train, train_idx)
val_dataset = Subset(full_dataset_val_test, val_idx)
test_dataset = Subset(full_dataset_val_test, test_idx)

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

# 5. Ajustar os modelos para o número correto de classes
model_resnet.fc = nn.Linear(model_resnet.fc.in_features, num_classes).to(device)
model_efficientnet.classifier[1] = nn.Linear(model_efficientnet.classifier[1].in_features, num_classes).to(device)

print("Caminhos e modelos corrigidos com sucesso!")
```

```python
import torch.optim as optim
import time
import copy


criterion = nn.CrossEntropyLoss()


optimizer_resnet = optim.Adam(model_resnet.parameters(), lr=0.001)
optimizer_efficient = optim.Adam(model_efficientnet.parameters(), lr=0.001)


def train_model(model, optimizer, num_epochs=5):
    since = time.time()
    best_acc = 0.0
    # Guarda os melhores pesos do modelo
    best_model_wts = copy.deepcopy(model.state_dict())

    for epoch in range(num_epochs):
        print(f'Época {epoch+1}/{num_epochs}')
        print('-' * 10)

        # Cada época tem uma fase de treino e uma de validação
        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()  # Modo de treino (aplica Data Augmentation, Dropout, etc.)
                dataloader = train_loader
            else:
                model.eval()   # Modo de avaliação (desliga os comportamentos de treino)
                dataloader = val_loader

            running_loss = 0.0
            running_corrects = 0

            # Itera sobre os lotes de imagens
            for inputs, labels in dataloader:
                inputs = inputs.to(device)
                labels = labels.to(device)

                # Zera os gradientes do otimizador
                optimizer.zero_grad()

                # Forward (passa a imagem na rede)
                # Só calcula os gradientes se estiver na fase de treino
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    # Backward + Otimização (atualiza os pesos)
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                # Estatísticas
                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            epoch_loss = running_loss / len(dataloader.dataset)
            epoch_acc = running_corrects.double() / len(dataloader.dataset)

            print(f'{phase.capitalize()} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            # Se for a melhor época na validação, salvamos o modelo
            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())
        print()

    time_elapsed = time.time() - since
    print(f'Treinamento completo em {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Melhor Acurácia de Validação: {best_acc:.4f}')

    # Carrega os melhores pesos antes de devolver o modelo
    model.load_state_dict(best_model_wts)
    return model

print("Função de treinamento pronta para ser usada!")
```

```python
print("Iniciando o treinamento da ResNet18...")
model_resnet = train_model(model_resnet, optimizer_resnet, num_epochs=5)
```

```python
print("Iniciando o treinamento da EfficientNet-B0...")
model_efficientnet = train_model(model_efficientnet, optimizer_efficient, num_epochs=5)
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report


def evaluate_model(model, dataloader, class_names, model_name):
    model.eval() # Modo de avaliação
    all_preds = []
    all_labels = []

    # Desliga o cálculo de gradientes para economizar memória e acelerar
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)

            # Move os tensores de volta para a CPU para usar no scikit-learn
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())


    print(f"================ Avaliação: {model_name} ================\n")
    print(classification_report(all_labels, all_preds, target_names=class_names))


    cm = confusion_matrix(all_labels, all_preds)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
    plt.ylabel('Rótulo Verdadeiro (Real)')
    plt.xlabel('Rótulo Predito (O que a rede achou)')
    plt.title(f'Matriz de Confusão - {model_name}')
    plt.xticks(rotation=45)
    plt.show()


evaluate_model(model_resnet, test_loader, class_names, "ResNet18")
evaluate_model(model_efficientnet, test_loader, class_names, "EfficientNet-B0")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import torchvision

# Função para desnormalizar a imagem e mostrá-la corretamente
def imshow(inp, title=None):
    inp = inp.numpy().transpose((1, 2, 0))
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    inp = std * inp + mean
    inp = np.clip(inp, 0, 1)
    plt.imshow(inp)
    if title is not None:
        plt.title(title)
    plt.pause(0.001)

# Função para buscar acertos e erros e plotá-los
def visualizar_qualitativo(model, dataloader, class_names, num_images=4):
    model.eval()
    images_so_far = 0
    fig = plt.figure(figsize=(12, 10))

    corretas = []
    incorretas = []

    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)

            # Separa as imagens corretas e incorretas
            for i in range(inputs.size()[0]):
                if preds[i] == labels[i] and len(corretas) < num_images:
                    corretas.append((inputs.cpu().data[i], preds[i], labels[i]))
                elif preds[i] != labels[i] and len(incorretas) < num_images:
                    incorretas.append((inputs.cpu().data[i], preds[i], labels[i]))

                if len(corretas) == num_images and len(incorretas) == num_images:
                    break
            if len(corretas) == num_images and len(incorretas) == num_images:
                break

    # Plota as corretas
    print("--- EXEMPLOS DE ACERTOS ---")
    plt.figure(figsize=(15, 4))
    for i, (img, pred, label) in enumerate(corretas):
        ax = plt.subplot(1, num_images, i + 1)
        ax.axis('off')
        ax.set_title(f'Previsto: {class_names[pred]}\nReal: {class_names[label]}', color='green')
        imshow(img)
    plt.show()

    # Plota as incorretas
    print("--- EXEMPLOS DE ERROS ---")
    plt.figure(figsize=(15, 4))
    for i, (img, pred, label) in enumerate(incorretas):
        ax = plt.subplot(1, num_images, i + 1)
        ax.axis('off')
        ax.set_title(f'Previsto: {class_names[pred]}\nReal: {class_names[label]}', color='red')
        imshow(img)
    plt.show()

# Rodando a análise qualitativa para o melhor modelo (você pode escolher qual visualizar)
print("Análise Qualitativa - ResNet18")
visualizar_qualitativo(model_resnet, test_loader, class_names)
```
